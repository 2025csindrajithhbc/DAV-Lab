```python
df = pd.read_csv("Iris.data.csv")
```

```python
print(df)
```

```python
plt.figure(figsize = (8, 5))
sns.boxplot(y = df['5.1'])
plt.title("Box plot of Plants")
plt.ylabel("Sepal length")
plt.show()
```

```python
Q1 = df['5.1'].quantile(0.25)
```

```python
Q3 = df['5.1'].quantile(0.75)
```

```python
IQR = Q3 - Q1
```

```python
print("Q1 = ",Q1)
print('Q3 = ',Q3)
print("IQR = ",IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
```

```python
print("Lower limit = ",lower_limit)
print("Upper limit = ",upper_limit)
```

```python
outliers_iqr = df[
    (df['5.1'] < lower_limit) | 
    (df['5.1'] > upper_limit)
    ]
```

```python
print("Outliers detected using IQR")
print(outliers_iqr)
```

```python
df["5.1_IQR_Capping"] = df['5.1'].clip(
    lower = lower_limit,
    upper = upper_limit
)
```
