```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pd.read_csv("iris.csv")
```

```python
df = pd.read_csv("iris.csv") 
```

```python
df
```

```python
sns.histplot(data = df , x = "petal_length",bins=10)
```

```python
plt.hist(df["petal_length"],bins=10)
plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.boxplot(df["petal_length"])
plt.title("Box Plot")
plt.show()
```

```python
sns.boxplot(data=df)
```

```python
df.boxplot(column="petal_length")
plt.show()
```

```python
df.boxplot(column="sepal_length")
plt.show()
```

```python
df.boxplot(column="sepal_width")
plt.show()
```

```python
sns.scatterplot(
    data = df,
    x = "petal_length",
    y = "petal_width",
    hue = "species")
```

```python
plt.scatter(df["petal_length"], df["petal_width"])
plt.titel("Petal Length vs Petal Width")
```
